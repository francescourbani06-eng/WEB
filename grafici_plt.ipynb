```python
import matplotlib.pyplot as plt
```

GRAFICO A LINEE

```python
x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 20, 25]
```

```python
plt.plot(x, y)

plt.title("Andamento dei valori")
plt.xlabel("Giorno")
plt.ylabel("Valore")
plt.plot(x, y, marker="o")
```

GRAFICO A BARRE

```python
nomi = ["Anna", "Luca", "Marco"]
voti = [25, 28, 30]
```

```python
plt.bar(nomi, voti)

plt.title("Voti degli studenti")
plt.xlabel("Studenti")
plt.ylabel("Voto")

plt.show()
```

ISTOGRAMMA

```python
voti = [18, 20, 22, 22, 23, 24, 24, 24, 25, 27, 28, 30]
```

```python
plt.hist(voti)

plt.title("Distribuzione dei voti")
plt.xlabel("Voto")
plt.ylabel("Frequenza")

plt.show()
```

SCATTER PLOT

```python
altezza = [160, 165, 170, 175, 180]
peso = [55, 60, 68, 72, 82]
```

```python
plt.scatter(altezza, peso)

plt.title("Altezza e peso")
plt.xlabel("Altezza (cm)")
plt.ylabel("Peso (kg)")

plt.show()
```
