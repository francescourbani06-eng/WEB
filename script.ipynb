```python
import os
import anthropic
from dotenv import load_dotenv
from anthropic import Anthropic
```

Carica la chiave API dal file .env

```python
load_dotenv()
```

Crea il client per comunicare con Claude

```python
client = Anthropic(
    api_key=os.getenv("ANTHROPIC_API_KEY")
)
```

Leggo dal file

```python
try:
    with open("script.txt", "r", encoding="utf-8") as file:
        testo = file.read()
except FileNotFoundError:
    print("Errore: Il file 'script.txt' non è stato trovato.")
    testo = ""
```

Menu con le opzioni e richiesta all'API

```python
if testo.strip():
    while True:
        # Mostra le opzioni all'utente
        print("\n--- MENU PRINCIPALE ---")
        print("1. Riassumi")
        print("2. Traduci")
        print("3. Esci")
        
        scelta = input("Seleziona un'opzione (1/2/3): ").strip()


        # Configura il prompt in base alla scelta
        if scelta == "1":
            prompt = f"Riassumi il seguente testo in poche righe evidenziando i punti chiave:\n\n{testo}"
        elif scelta == "2":
            prompt = f"Traduci il seguente testo mantenendo uno stile fluido e naturale:\n\n{testo}"
        elif scelta == "3":
            print("Chiusura del programma. Arrivederci!")
            break
        else:
            print("Scelta non valida. Inserisci 1, 2 o 3.")
            continue  # Torna all'inizio del ciclo


        # 3. Chiamata API protetta da try/except
        try:
            print("\nInvio della richiesta a Claude...")
            
            risposta = client.messages.create(
                model="claude-3-5-haiku-20241022",
                max_tokens=500,
                messages=[
                    {"role": "user", "content": prompt}
                ]
            )
            
            # Stampa il risultato
            print("\n--- RISULTATO ---")
            for blocco in risposta.content:
                if blocco.type == "text":
                    print(blocco.text)

        except anthropic.APIConnectionError:
            print("\n[Errore] Impossibile connettersi ai server. Controlla la tua connessione a Internet.")
        except anthropic.AuthenticationError:
            print("\n[Errore] Chiave API non valida o mancante. Verifica il file .env.")
        except anthropic.APIError as e:
            print(f"\n[Errore API] Si è verificato un problema con la chiamata: {e.message}")
        except Exception as e:
            print(f"\n[Errore imprevisto] {e}")

else:
    print("Impossibile proseguire: il file 'script.txt' è vuoto o mancante. Inserisci del testo nel file e riavvia lo script.")
```
